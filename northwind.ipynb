```python
# convert orderDate into date datatype 
```

```python
orders['orderDate']=pd.to_datetime(orders['orderDate'],format='%d-%m-%Y')
orders['requiredDate']=pd.to_datetime(orders['requiredDate'],format='%d-%m-%Y')
orders['shippedDate']=pd.to_datetime(orders['shippedDate'],format='%d-%m-%Y')
```

```python
orders.info()
```

```python
(~orders['customerID'].isin(customers['customerID'])).sum()
(~orders['employeeID'].isin(employees['employeeID'])).sum()
(~orders['shipperID'].isin(shippers['shipperID'])).sum()
(~order_details['orderID'].isin(orders['orderID'])).sum()
(~order_details['productID'].isin(products['productID'])).sum()
(~products['categoryID'].isin(categories['categoryID'])).sum()
```

```python
orders['orderYear']=orders['orderDate'].dt.year
orders['orderMonth']=orders['orderDate'].dt.month
orders['daystoship']=(orders['shippedDate']-orders['orderDate']).dt.days
orders['islate']=orders['shippedDate']>orders['requiredDate']
```

```python
orders[['orderDate', 'orderYear', 'orderMonth', 'daystoship', 'islate']].head()
orders['islate'].sum()
```

```python
Q1 = orders['shippingcost'].quantile(0.25)
Q3 = orders['shippingcost'].quantile(0.75)
upper_limit = Q3 + 1.5 * (Q3 - Q1)

upper_limit
(orders['shippingcost'] > upper_limit).sum()
```

```python
# turn column names into lowercase
```

```python
for table in [shippers, employees, categories, products, customers, order_details, orders]:
    table.columns = table.columns.str.lower()
```

```python
shippers.to_csv('clean_shippers.csv', index=False)
employees.to_csv('clean_employees.csv', index=False)
categories.to_csv('clean_categories.csv', index=False)
products.to_csv('clean_products.csv', index=False, encoding='utf-8')
customers.to_csv('clean_customers.csv', index=False, encoding='utf-8')
order_details.to_csv('clean_order_details.csv', index=False)
orders.to_csv('clean_orders.csv', index=False)
```

```python
from sqlalchemy import create_engine
from getpass import getpass

username = "postgres"
password = getpass("PostgreSQL password: ")
host = "localhost"
port = "5432"
database = "northwind"

engine = create_engine(f"postgresql+psycopg2://{username}:{password}@{host}:{port}/{database}")

shippers.to_sql("shippers", engine, if_exists="replace", index=False)
employees.to_sql("employees", engine, if_exists="replace", index=False)
categories.to_sql("categories", engine, if_exists="replace", index=False)
products.to_sql("products", engine, if_exists="replace", index=False)
customers.to_sql("customers", engine, if_exists="replace", index=False)
order_details.to_sql("order_details", engine, if_exists="replace", index=False)
orders.to_sql("orders", engine, if_exists="replace", index=False)

print("All 7 tables loaded into", database)
```
